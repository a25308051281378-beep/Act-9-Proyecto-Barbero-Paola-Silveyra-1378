#### 55. Barbero
#### Usando LISTAS + if + elif + else + while True

```python
# ==========================================
# PROYECTO UIII ACT 9
# IA BARBERO CON LISTAS
# ==========================================

print("======================================")
print("💈 IA INTELIGENTE DE BARBERO 💈")
print("======================================")

# Lista vacía
clientes = []

while True:

    print("\n=========== MENÚ ===========")
    print("1. Agregar cliente")
    print("2. Ver clientes")
    print("3. Recomendación de corte")
    print("4. Eliminar cliente")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        nombre = input("\nIngresa el nombre del cliente: ")

        clientes.append(nombre)

        print("✅ Cliente agregado correctamente")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n📋 LISTA DE CLIENTES")

        if clientes == []:
            print("❌ No hay clientes registrados")

        else:
            print(clientes)

    # OPCIÓN 3
    elif opcion == "3":

        print("\n✂ RECOMENDACIÓN DE CORTE")

        estilo = input("¿Te gusta un estilo moderno? (si/no): ")

        if estilo == "si":
            print("✅ IA recomienda corte Fade")

        else:
            print("✅ IA recomienda corte Clásico")

    # OPCIÓN 4
    elif opcion == "4":

        eliminar = input("\nNombre del cliente a eliminar: ")

        if eliminar in clientes:

            clientes.remove(eliminar)

            print("✅ Cliente eliminado")

        else:
            print("❌ Cliente no encontrado")

    # OPCIÓN 5
    elif opcion == "5":

        print("\n👋 Gracias por usar la IA de Barbero")
        break

    # ERROR
    else:
        print("\n❌ Opción inválida")
        print("Programa desarrollado por Paola Silveyra NL = 55")
```
